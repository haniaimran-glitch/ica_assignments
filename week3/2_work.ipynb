```python
#further cleaning my week 2 file before moving on 
clean_1['item']=clean_1['item'].str.strip().str.title()
clean_1['date'] = pd.to_datetime(clean_1["date"])
clean_1.info()
clean_1['item'].value_counts()
```

using original clean data

```python
clean.groupby('item')['price'].agg(['mean','median']).round(2)
```

```python
# Arranging the data according to date
clean_sorted = clean.sort_values('date')
# Grouping the data by item and finding first prices
first_prices_org = clean_sorted.groupby('item')['price'].first()
# Grouping the data by item and finding last prices
last_prices_org = clean_sorted.groupby('item')['price'].last()
# Creating a table of first and last prices
first_last = pd.DataFrame({'first': first_prices_org,'last': last_prices_org})
# Calculating percentage change
first_last['percentage_change'] = ((first_last['last'] - first_last['first'])/ first_last['first'] * 100).round(2)
# Printing the results
first_last
```

using my clean data WITH THE FLAGGED ITEM

```python
#with flagged data 
# Finding the mean and median for each item with flagged data
clean_1.groupby('item')['price'].agg(['mean', 'median'])
```

```python
# Arranging the data according to date
clean_1_sorted = clean_1.sort_values('date')
# Grouping the data by item and finding first prices
first_prices = clean_1_sorted.groupby('item')['price'].first()
# Grouping the data by item and finding last prices
last_prices = clean_1_sorted.groupby('item')['price'].last()
# Creating a table of first and last prices
first_last_1 = pd.DataFrame({'first_1': first_prices,'last_1': last_prices})
# Calculating percentage change
first_last_1['percentage_change'] = ((first_last_1['last_1'] - first_last_1['first_1'])/ first_last_1['first_1'] * 100).round(2)
# Printing the results
first_last_1
```

using my clean data WITHOUT THE FLAGGED ITEM 

```python
#FINDING THE OUTLIER
clean_1['price'].describe()
```

```python
# The descriptive statistics show that the maximum price (9999), which is significantly higher than the other observations and since 75% values are under 302.5 , making 9999 a potential outlier.
#finding the row of the outlier
clean_1[clean_1['price'] == 9999]
```

```python
# Removing the flagged value(outlier) from row 30 of the data
clean_1_without_outlier = clean_1.drop(30, axis=0)
```

```python
#Finding the mean and median for each item without flagged data
clean_1_sorted_without_outlier.groupby('item')['price'].agg(['mean', 'median'])
```

```python
# Arranging the data according to date
clean_1_sorted_without_outlier = clean_1_without_outlier.sort_values('date')
# Grouping the data by item and finding first and last prices
first_prices_without = clean_1_sorted_without_outlier.groupby('item')['price'].first()
last_prices_without = clean_1_sorted_without_outlier.groupby('item')['price'].last()
# Creating a table of first and last prices
first_last_without = pd.DataFrame({'first_price': first_prices_without,'last_price': last_prices_without})
# Calculating percentage change between first and last prices
first_last_without['percentage_change'] = ((first_last_without['last_price'] - first_last_without['first_price'])/ first_last_without['first_price'] * 100).round(2)
# Printing the results
first_last_without
```
