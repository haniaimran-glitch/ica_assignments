AI CLEANING CODE

My prompt to AI :I have a teaching dataset containing 97 rows and 6 columns: date, item, city, price, unit, and data_label.
All six columns currently have a string data type. The dataset contains one missing value in the price column, located in row 5. It also contains one exact duplicate observation, where row 96 is identical to row 8.
The date column needs to be converted into datetime format, while the price column needs to be converted into a numeric data type for further calculations and analysis.
Please suggest Python cleaning code using Pandas to address these issues. The missing price should remain missing rather than being filled with an assumed value. The exact duplicate should be removed only from a working copy of the dataset, while the original dataset must remain unchanged.
Please explain each line of code.

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Convert the date column into datetime format
df_clean['date'] = pd.to_datetime(df_clean['date'])

# Convert the price column into numeric format
df_clean['price'] = pd.to_numeric(df_clean['price'], errors="coerce")

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()
```

```python
# Check the data types and non-null values
df_clean.info()

# Check missing values
df_clean.isna().sum()

# Check remaining duplicates
df_clean.duplicated().sum()

# Check the updated number of rows and columns
df_clean.shape
```
